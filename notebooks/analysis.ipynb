# Hospital Performance Analytics

## Exploratory Data Analysis using the MIMIC-IV Demo Dataset

This notebook demonstrates how SQL, SQLite, Pandas and Matplotlib can be used to explore hospital data and visualize important healthcare insights.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

## Connect to SQLite Database

```python
conn = sqlite3.connect("../data/Hospital-Performance-Analytics_hospital.db")
```

## Load Hospital Tables

```python
patients = pd.read_sql_query(
    "SELECT * FROM patients;",
    conn
)

admissions = pd.read_sql_query(
    "SELECT * FROM admissions;",
    conn
)

services = pd.read_sql_query(
    "SELECT * FROM services;",
    conn
)
```

## Preview Patients Dataset

```python
patients.head()
patients.info()
patients.describe()
```

## Gender Distribution

```python
patients["gender"].value_counts().plot(
    kind="bar"
)

plt.title("Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

## Age Distribution

```python
patients["anchor_age"].plot(
    kind="hist",
    bins=15
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

## Admission Type Distribution

```python
admissions["admission_type"].value_counts().plot(
    kind="bar"
)

plt.title("Admissions by Type")
plt.show()
```

# Insurance Distribution

```python
admissions["insurance"].value_counts().plot(
    kind="bar"
)

plt.title("Insurance Distribution")
plt.show()
```

# Race Distribution

```python
admissions["race"].value_counts().head(10).plot(
    kind="bar"
)

plt.title("Race Distribution")
plt.show()
```

## Current Hospital Services

```python
services["curr_service"].value_counts().plot(
    kind="bar"
)

plt.title("Current Hospital Services")
plt.show()
```

## Previous Hospital Services

```python
services["prev_service"].value_counts().head(10).plot(
    kind="bar"
)

plt.title("Top Previous Hospital Services")
plt.show()
```

## Admissions per Year

```python
admissions["admittime"] = pd.to_datetime(
    admissions["admittime"]
)

admissions["year"] = admissions["admittime"].dt.year

admissions["year"].value_counts().sort_index().plot(
    kind="bar"
)

plt.title("Admissions per Year")
plt.show()
```

## Close Database

```python
conn.close()
```
